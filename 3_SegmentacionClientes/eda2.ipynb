```python
##### Investigando 'NUM_VISITANTES' por 'TEMPORADA'
#### Creando 'TEMPORADA'
### Seleccionando datos
df_amn = df_1[['ANIO', 'ID_MES','NUMERO_VISITANTES']]\
    .groupby(by=['ANIO', 'ID_MES'], as_index=False)\
    ['NUMERO_VISITANTES'].sum()
# display(df_amn)
### Creando mascaras
mask_verano =  (df_amn['ID_MES']== 1)| (df_amn['ID_MES']== 2) | (df_amn['ID_MES']== 3)
mask_otono=  (df_amn['ID_MES']== 4)| (df_amn['ID_MES']== 5) | (df_amn['ID_MES']== 6)
mask_invierno=  (df_amn['ID_MES']== 7)| (df_amn['ID_MES']== 8) | (df_amn['ID_MES']== 9)
mask_primavera=  (df_amn['ID_MES']== 10)| (df_amn['ID_MES']== 11) | (df_amn['ID_MES']== 12)
### Creando columna 'TEMPORADA'
ds_temporada = pd.Series(np.nan, index=df_amn.index, dtype=str)
ds_temporada.name = 'TEMPORADA'
ds_temporada.iloc[df_amn[mask_verano].index] = 'Verano'
ds_temporada.iloc[df_amn[mask_otono].index] = 'Otono'
ds_temporada.iloc[df_amn[mask_invierno].index] = 'Invierno'
ds_temporada.iloc[df_amn[mask_primavera].index] = 'Primavera'
### Concat
df_amn = pd.concat([df_amn, ds_temporada], axis=1)
# display(df_amn)


#### Filtrando 'NUMERO_VISITANTES' basado en 'TEMPORADA'
### Creando nuevo df
df_atn = df_amn[['ANIO', 'TEMPORADA', 'NUMERO_VISITANTES']]\
    .groupby(by=['ANIO', 'TEMPORADA'], as_index=False)\
    ['NUMERO_VISITANTES'].mean().round(2)
display(df_atn)
### Configuraciones de variables
anios = df_atn['ANIO'].value_counts().index.to_list()
temporadas = df_atn['TEMPORADA'].value_counts().index.to_list()
### Creando line graph
ANCH = 8
ALT = 5
plt.figure(figsize=(ANCH, ALT))
colores = ['red', 'blue', 'green', 'turquoise']
for i in range(len(temporadas)):
    plt.plot(anios,
            df_atn[df_atn['TEMPORADA']==temporadas[i]]['NUMERO_VISITANTES'],
            label = temporadas[i],
            color = colores[i],
            marker='o'
    )
plt.title('Visitantes por Temporada entre 2019 - 2025')
plt.xlabel('Anio')
plt.ylabel('Numero de Visitantes')
plt.grid(True,axis='y', linestyle='--', alpha=0.9)
plt.legend(fontsize=ANCH, title='Temporada')
plt.show()
```
